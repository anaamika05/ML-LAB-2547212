Task 2: Train-Test Split Analysis

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

Create a function for different train-test splits

```python
def evaluate_split(test_size):

    # Split the data
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        random_state=42
    )

    # Feature Scaling
    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # KNN Model (using K = 5 for now)
    knn = KNeighborsClassifier(n_neighbors=5)

    # Train the model
    knn.fit(X_train_scaled, y_train)

    # Make predictions
    y_pred = knn.predict(X_test_scaled)

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)

    print(f"Train-Test Split: {int((1-test_size)*100)}:{int(test_size*100)}")
    print("Training Samples:", len(X_train))
    print("Testing Samples :", len(X_test))
    print("Accuracy:", round(accuracy, 4))
    print("-" * 40)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Use 80:20 split for the remaining tasks
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Test all three splits

```python
evaluate_split(0.20)   # 80:20

evaluate_split(0.30)   # 70:30

evaluate_split(0.10)   # 90:10
```

Inference (Task 2: Train-Test Split Analysis)
The model was evaluated using three train-test splits: 80:20, 70:30, and 90:10.
Increasing the training data generally helps the model learn better patterns, which can improve prediction performance.
The 80:20 split provides a good balance between training and testing data, making it a commonly preferred choice.
The 70:30 split has more testing data, which gives a more reliable estimate of the model's performance but slightly reduces the amount of training data.
The 90:10 split provides the largest training set, but the small test set may not represent the entire dataset well, making the evaluation less stable.
Overall, different train-test splits may produce slightly different accuracy values, showing that model performance depends on how the data is divided. A balanced split helps achieve better generalization and a more reliable evaluation.

Task 3.1: Heuristic Method for K Selection

```python
import math

# Number of training samples
n = len(X_train)

# Heuristic K value
k = int(math.sqrt(n))

# Make K odd (recommended for binary classification)
if k % 2 == 0:
    k += 1

print("Number of training samples:", n)
print("Heuristic K value:", k)
```

Task 3.2: Model Training

```python
k_values = [k-5, k-3, k-1, k, k+1, k+3, k+5]

accuracy_list = []

for i in k_values:
    
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train_scaled, y_train)

    y_pred = knn.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)

    accuracy_list.append(accuracy)

    print(f"K = {i}  Accuracy = {accuracy:.4f}")
```

Plot Accuracy vs K

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(k_values, accuracy_list, marker='o')
plt.title("Accuracy vs K")
plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.grid(True)
plt.show()
```

Find the Best K

```python
best_k = k_values[accuracy_list.index(max(accuracy_list))]

print("Best K:", best_k)
print("Highest Accuracy:", max(accuracy_list))
```

Task 3.3: Distance Metrics and Decision Boundary Mapping.

```python
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
import numpy as np
```

Reduce the Dataset to 2 Dimensions

```python
# Apply PCA
pca = PCA(n_components=2)

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
```

Function to Plot Decision Boundary

```python
def plot_decision_boundary(X, y, k):

    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X, y)

    h = 0.02

    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, h),
        np.arange(y_min, y_max, h)
    )

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(7,6))

    plt.contourf(xx, yy, Z, alpha=0.3)

    plt.scatter(
        X[:,0],
        X[:,1],
        c=y,
        edgecolors='k'
    )

    plt.title(f"KNN Decision Boundary (K={k})")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")

    plt.show()
```

Plot for Different K Values

```python
plot_decision_boundary(X_train_pca, y_train, 1)

plot_decision_boundary(X_train_pca, y_train, 5)

plot_decision_boundary(X_train_pca, y_train, 10)

plot_decision_boundary(X_train_pca, y_train, 20)
```

Task 4: Cross Validation

Import Required Libraries

```python
from sklearn.model_selection import cross_val_score
```

Test Different K Values

```python
k_values = [17, 19, 21, 23, 25]

mean_accuracy = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)

    scores = cross_val_score(
        knn,
        X_scaled,
        y,
        cv=5,
        scoring='accuracy'
    )

    mean_accuracy.append(scores.mean())

    print(f"K = {k}")
    print("Fold Accuracies:", scores)
    print("Mean Accuracy:", round(scores.mean(), 4))
    print("-" * 40)
```

Plot Mean Accuracy vs K

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(k_values, mean_accuracy, marker='o')

plt.title("Cross Validation Accuracy vs K")
plt.xlabel("K Value")
plt.ylabel("Mean Accuracy")
plt.grid(True)

plt.show()
```

Find the Best K

```python
best_k = k_values[mean_accuracy.index(max(mean_accuracy))]

print("Best K:", best_k)
print("Highest Mean Accuracy:", round(max(mean_accuracy), 4))
```

The KNN model was evaluated using 5-fold cross-validation for different K values. The mean accuracy obtained from cross-validation provides a more reliable estimate of model performance than a single train-test split because every sample is used for both training and testing across different folds. The K value with the highest mean accuracy was selected as the optimal value, ensuring better generalization and reducing the effect of random data splitting.

Task 6: Comparative Study with Regression

### 1. Compare regression evaluation metrics with classification metrics.

Regression and classification use different evaluation metrics because they solve different types of problems. Regression predicts continuous numerical values and uses metrics such as **MAE, MSE, RMSE, and R² Score** to measure prediction errors. Classification predicts categories or classes and uses metrics such as **Accuracy, Precision, Recall, F1 Score, Confusion Matrix, and ROC-AUC** to measure how correctly the model classifies data.

### 2. Explain the difference between Error-based Evaluation (Regression) and Decision-based Evaluation (Classification).

**Error-based evaluation** is used in regression to measure how close the predicted values are to the actual values. Lower error values indicate better model performance.

**Decision-based evaluation** is used in classification to measure how accurately the model assigns data to the correct class. It evaluates the correctness of the model's decisions rather than numerical prediction errors.

### 3. Compare R² Score and Accuracy.

**R² Score** is used in regression and measures how well the model explains the variation in the target variable. A value closer to 1 indicates a better fit.

**Accuracy** is used in classification and measures the proportion of correctly classified instances out of the total number of predictions. Higher accuracy indicates better classification performance.

### 4. Compare RMSE and F1 Score.

**RMSE** measures the average magnitude of prediction errors in regression. A lower RMSE indicates that the predicted values are closer to the actual values.

**F1 Score** is used in classification and combines Precision and Recall into a single metric. A higher F1 Score indicates a better balance between correctly identifying positive cases and minimizing false predictions.

### 5. Compare MAE and Confusion Matrix.

**MAE (Mean Absolute Error)** measures the average absolute difference between predicted and actual values in regression. Smaller MAE values indicate better predictions.

A **Confusion Matrix** is used in classification to show the number of correctly and incorrectly classified instances. It includes True Positives, True Negatives, False Positives, and False Negatives, helping analyze the model's classification performance.

### 6. Discuss the differences in evaluation logic for Continuous Prediction Tasks and Classification Tasks.

In **continuous prediction tasks**, the objective is to predict numerical values as accurately as possible. Therefore, evaluation focuses on measuring prediction errors using metrics like MAE, MSE, RMSE, and R² Score.

In **classification tasks**, the objective is to correctly assign data to predefined classes. Therefore, evaluation focuses on decision correctness using metrics such as Accuracy, Precision, Recall, F1 Score, Confusion Matrix, and ROC-AUC.



## **Task 7: Analytical Questions**

### **1. Why is KNN called a lazy learning algorithm?**

KNN is called a **lazy learning algorithm** because it does not build a model during the training phase. Instead, it stores the training data and performs the classification only when a new data point needs to be predicted.

### **2. Why is feature scaling required in KNN?**

Feature scaling is required because KNN uses distance calculations to classify data. If one feature has much larger values than others, it will dominate the distance calculation. StandardScaler ensures all features have a similar scale, allowing each feature to contribute equally.

### **3. Explain heuristic K selection using the √n rule.**

The heuristic method selects the initial value of K using the formula:

[
K = \sqrt{n}
]

where **n** is the number of training samples. The calculated value is usually rounded to the nearest odd number to avoid ties in binary classification. This provides a reasonable starting point for selecting the optimal K.

### **4. Why is cross-validation more reliable than a single train-test split?**

Cross-validation is more reliable because the dataset is divided into multiple folds, and each fold is used as both training and testing data at different times. This reduces the effect of random data splitting and provides a more accurate estimate of the model's performance.

### **5. How does K affect the bias-variance trade-off?**

A **small K** value results in low bias but high variance, making the model sensitive to noise and more likely to overfit. A **large K** value results in high bias but low variance, making the model simpler and more likely to underfit. Choosing an appropriate K provides a balance between bias and variance.

### **6. Why is recall more important than accuracy in cancer prediction?**

Recall is more important because it measures how many actual cancer cases are correctly identified by the model. Missing a cancer patient (false negative) can delay diagnosis and treatment, which can have serious consequences. Therefore, maximizing recall is more important than achieving high overall accuracy.

### **7. What is the limitation of very large K values?**

Very large K values consider too many neighboring data points, causing the model to ignore important local patterns. This leads to over-smoothing of the decision boundary, increased bias, and reduced classification accuracy, especially when different classes are close together.

The regression metrics used in **Lab 3**, such as **MAE, MSE, RMSE, and R² Score**, evaluate the magnitude of prediction errors by measuring the difference between the predicted and actual numerical values. Lower values of MAE, MSE, and RMSE indicate better prediction accuracy, while a higher R² Score indicates that the model explains a larger proportion of the variance in the data.

In **Lab 4**, classification metrics evaluate how correctly the model classifies data into different categories. Metrics such as **Accuracy, Precision, Recall, F1 Score, Confusion Matrix, and ROC-AUC** measure the correctness of classification decisions rather than numerical prediction errors.

Accuracy alone is not sufficient for medical diagnosis because a model may achieve high accuracy while still failing to identify patients with the disease. In healthcare applications such as breast cancer detection, **Recall** is more important because it measures the model's ability to correctly identify actual positive cases, reducing the risk of missed diagnoses. **ROC-AUC** is also an important metric because it evaluates the model's ability to distinguish between malignant and benign cases across different classification thresholds.

Overall, regression and classification use different evaluation frameworks based on the nature of the prediction task. Regression focuses on minimizing prediction errors for continuous values, whereas classification focuses on maximizing the correctness of decisions and identifying the appropriate class labels accurately.
